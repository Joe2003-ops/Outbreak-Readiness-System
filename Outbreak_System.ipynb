```python
import requests
import pandas as pd

# Download HDX Kenya Healthsites dataset directly
url = "https://data.humdata.org/dataset/df79db22-da00-4c05-bda0-79c14f27132d/resource/4d2047ed-5dd8-4e8c-85f0-622fbc789b53/download/kenya-healthsites.csv"
df_facilities = pd.read_csv(url)

# Preview initial structure
print(df_facilities[['name', 'amenity', 'latitude', 'longitude']].head())
```

```python
import geopandas as gpd

# Load county boundaries
counties_gdf = gpd.read_file("data/raw/kenya_county_boundaries.geojson")

# Convert facility coordinates into Geospatial Point geometries
facilities_gdf = gpd.GeoDataFrame(
    df_facilities, 
    geometry=gpd.points_from_xy(df_facilities.longitude, df_facilities.latitude),
    crs="EPSG:4326"
)

# Spatial Join: Assign each health facility to its county automatically
facilities_with_county = gpd.sjoin(facilities_gdf, counties_gdf, how="inner", predicate="intersects")
```
